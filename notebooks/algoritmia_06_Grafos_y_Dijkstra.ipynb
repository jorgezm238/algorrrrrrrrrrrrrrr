# 5. Dijkstra: la idea

Dijkstra mantiene:

- `distancias`: mejor distancia conocida desde el origen;
- `visitados`: nodos cuya distancia consideramos definitiva.

Al inicio desde `A`:

```text
distancias = [0, ∞, ∞, ∞, ∞]
visitados  = [F, F, F, F, F]
```

## Decisión voraz

\[
\boxed{\text{elegir el nodo NO visitado con menor distancia conocida}}
\]


## 5.1. Pseudocódigo

```text
ALGORITMO Dijkstra(grafo, origen)

    distancia[origen] ← 0

    PARA cada nodo distinto del origen
        distancia[nodo] ← ∞
    FIN PARA

    visitado[nodo] ← FALSO para todos

    REPETIR V veces

        actual ← nodo no visitado
                  con menor distancia conocida

        marcar actual como visitado

        PARA cada posible vecino de actual

            SI existe arista entre actual y vecino
               Y vecino no está visitado

                nueva ← distancia[actual]
                         + peso(actual, vecino)

                SI nueva < distancia[vecino]
                    distancia[vecino] ← nueva
                FIN SI

            FIN SI

        FIN PARA

    FIN REPETIR

    DEVOLVER distancia
FIN ALGORITMO
```

Orden recomendado:

1. buscar el no visitado con menor distancia;

4. calcular la nueva distancia;
5. actualizar si mejora.

### Reconstrucción del camino

````python
def reconstruir_camino(predecesor, origen, destino):

    camino = []
    actual = destino

    while actual != -1:
        camino.append(actual)

        if actual == origen:
            break

        actual = predecesor[actual]

    camino.reverse()

    if not camino or camino[0] != origen:
        return []

    return camino
````

# 12. ¿Cuándo podemos confiar en Dijkstra?

Dijkstra marca como definitiva la distancia del nodo no visitado más cercano.

Esto requiere **pesos no negativos**.

Ejemplo:

```text
A --2--> B
A --5--> C
C ---10--> B
```

Inicialmente `B` parece estar a distancia 2, pero:

```text
A → C → B
5 + (-10) = -5
```

es mejor.

Por tanto:

\[
\boxed{\text{Dijkstra no debe utilizarse con pesos negativos}}
\]


# 13. ¿Cuánto trabajo hace nuestra versión?

Usamos una matriz de adyacencia.

En cada una de las `V` iteraciones:

- buscamos el mínimo entre hasta `V` nodos;
- recorremos hasta `V` posibles vecinos.

Por tanto:

\[
\boxed{O(V^2)}
\]
