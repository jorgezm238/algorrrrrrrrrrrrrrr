## 1. Variables, tipos y operaciones

Una **variable** asocia un nombre a un valor. La asignación `edad = 20` permite usar después el nombre `edad`. El signo `=` asigna; no expresa una igualdad matemática que se mantiene automáticamente.

No declaramos el tipo como en C. Los valores sí tienen tipo: `int` para enteros, `float` para números con decimales, `str` para textos y `bool` para los valores lógicos `True` y `False`. `type(valor)` permite consultar ese tipo. El separador decimal es el punto.

Los nombres distinguen mayúsculas y minúsculas: `total` y `Total` son diferentes. Usa nombres descriptivos, sin espacios, como `precio_final`.

| Operador | Significado | Ejemplo |
|:--|:--|:--|
| `+`, `-`, `*` | Suma, resta, producto | `3 * 4` → `12` |
| `/` | División | `7 / 2` → `3.5` |
| `//` | División redondeada hacia abajo | `7 // 2` → `3` |
| `%` | Resto | `7 % 2` → `1` |
| `**` | Potencia | `2 ** 3` → `8` |

Los paréntesis permiten agrupar operaciones. Para estas primeras actividades, asignaremos los datos directamente en el código.

**Ejemplo resuelto: calcular el coste de una compra**

````python
precio = 12.5
unidades = 4
total = precio * unidades
print("Total:", total)
print("Tipo de precio:", type(precio))
print("Tipo de unidades:", type(unidades))
````

## 2. Condicionales: tomar decisiones

Un condicional ejecuta un bloque cuando una condición es verdadera. Escribimos `if`, la condición y dos puntos. El bloque se marca con **indentación**: utiliza cuatro espacios, de forma consistente. Python no utiliza llaves para delimitarlo.

`elif` permite comprobar otra condición si las anteriores fueron falsas. `else` recoge el caso restante. En una cadena `if`–`elif`–`else` se ejecuta como máximo una rama.

| Comparación | Significado |
|:--|:--|
| `==`, `!=` | Igual, distinto |
| `<`, `<=` | Menor, menor o igual |
| `>`, `>=` | Mayor, mayor o igual |

Las comparaciones producen `True` o `False`. Para combinar condiciones usamos `and` (ambas), `or` (al menos una) y `not` (negación).

**Atención:** `=` asigna; `==` compara.

**Ejemplo resuelto: clasificar una temperatura**

````python
temperatura = 18

if temperatura < 10:
    print("Hace frío")
elif temperatura < 25:
    print("Temperatura templada")
else:
    print("Hace calor")

print("Clasificación terminada")
````

La última línea no está indentada: se ejecuta después del condicional, cualquiera que sea la rama elegida.

**Ejemplo resuelto: comprobar un intervalo**

````python
nota = 7.5
nota_valida = nota >= 0 and nota <= 10
print("¿Nota válida?", nota_valida)
````

## 3. Bucles: repetir instrucciones

### 3.1. `for` y `range`

`for` recorre los elementos de una secuencia o de otro objeto iterable. Para repetir con enteros utilizamos `range`:

- `range(fin)` produce los enteros desde 0 hasta `fin - 1`.
- `range(inicio, fin)` comienza en `inicio` y **excluye `fin`**.
- `range(inicio, fin, paso)` permite elegir el incremento; por ejemplo, `range(2, 9, 2)` recorre 2, 4, 6 y 8.

`range` no construye una lista con todos esos enteros. El cuerpo del bucle lleva dos puntos e indentación, como un condicional.

**Ejemplo resuelto: mostrar los cuadrados del 1 al 4**

````python
for numero in range(1, 5):
    cuadrado = numero ** 2
    print(numero, cuadrado)
````

### 3.2. Acumuladores

Un **acumulador** conserva un resultado parcial. Se inicializa antes del bucle y se actualiza dentro. Para una suma solemos comenzar en 0; para un producto, en 1.

`total += valor` abrevia `total = total + valor` en los cálculos numéricos que hacemos aquí.

**Ejemplo resuelto: sumar los cuadrados del 1 al 4**

````python
total = 0
for numero in range(1, 5):
    total += numero ** 2

print("Suma de cuadrados:", total)
````

### 3.3. `while`: repetir mientras se cumpla una condición

`while` comprueba su condición antes de cada vuelta. Si es falsa desde el principio, no ejecuta el cuerpo. Debes actualizar las variables necesarias para que el bucle pueda terminar; de lo contrario puedes crear un **bucle infinito**.

**Ejemplo resuelto: cuenta atrás**

````python
restantes = 3
while restantes > 0:
    print(restantes)
    restantes = restantes - 1

print("Fin")
````

## 4. Listas: trabajar con varios datos

Una **lista** es una colección ordenada y modificable. Se escribe entre corchetes, con elementos separados por comas: `[10, 20, 30]`. La lista vacía es `[]`.

Los índices empiezan en **0**: en una lista de longitud `n`, van de 0 a `n - 1`. `len(lista)` devuelve su longitud. Un índice fuera del rango produce `IndexError`.

- `lista[i]` consulta el elemento de índice `i`.
- `lista[i] = valor` modifica ese elemento.
- `lista.append(valor)` añade un elemento al final.
- `for elemento in lista` recorre directamente sus valores.
- `for i in range(len(lista))` recorre sus índices.

**Ejemplo resuelto: acceder y modificar**

````python
notas = [6, 8, 5]
print("Primera nota:", notas[0])
notas[2] = 7
notas.append(9)
print("Notas:", notas)
print("Cantidad:", len(notas))
````

**Ejemplo resuelto: recorrer valores y contar**

Un contador aumenta en uno cada vez que ocurre algo. Aquí contamos las notas mayores o iguales que 7.

````python
notas = [6, 8, 5, 9]
cantidad = 0

for nota in notas:
    if nota >= 7:
        cantidad += 1

print("Notas de al menos 7:", cantidad)
````

**Ejemplo resuelto: recorrer índices**

Usamos índices cuando necesitamos conocer la posición de cada elemento.

````python
colores = ["rojo", "verde", "azul"]
for i in range(len(colores)):
    print("Índice:", i, "Valor:", colores[i])
````

### Una diferencia importante: compartir no es copiar

Las listas son **mutables**: pueden cambiar. Con `otra = datos`, ambos nombres se refieren a la misma lista. Para crear una copia independiente de una lista de números podemos usar `datos.copy()`.

Un corte como `datos[1:3]` crea una nueva lista con los elementos de índices 1 y 2: el extremo final se excluye. Copiar `k` elementos requiere tiempo y espacio proporcionales a `k`; una expresión corta también puede tener coste.

Tanto `copy()` como los cortes hacen copias superficiales: si hubiera listas dentro, esas listas internas seguirían compartidas. Hoy trabajamos con listas de números.

**Ejemplo resuelto: observa qué cambia**

````python
datos = [10, 20, 30]
otra = datos
copia = datos.copy()
otra[0] = 99

print("Original:", datos)
print("Misma lista:", otra)
print("Copia:", copia)
print("Corte:", datos[1:3])
````

## 5. Funciones: poner nombre a una tarea

Una **función** agrupa instrucciones que podemos reutilizar. Se define con `def`, un nombre, los parámetros entre paréntesis y dos puntos. Su cuerpo va indentado.

Los **parámetros** son los nombres que recibe la función; los **argumentos** son los valores concretos que proporcionamos al llamarla. Las variables que asignamos dentro son normalmente locales a esa llamada.

`return` termina la llamada y devuelve un resultado que podemos guardar o utilizar en otro cálculo. **`print` muestra; `return` devuelve.** Una función que termina sin un `return` con valor devuelve `None`, que representa ausencia de valor.

**Ejemplo resuelto: calcular y reutilizar un resultado**

````python
def area_triangulo(base, altura):
    area = base * altura / 2
    return area

resultado = area_triangulo(6, 4)
print("Área:", resultado)
print("El doble del área:", 2 * resultado)
````

**Ejemplo resuelto: devolver un valor lógico**

Una comparación ya produce un booleano; podemos devolverlo directamente.

````python
def es_positivo(numero):
    return numero > 0

print(es_positivo(3))
print(es_positivo(-2))
print(es_positivo(0))
````

### Corrección y coste: explica tu algoritmo

Un **invariante** es una propiedad que se mantiene durante un proceso. Aquí, después de cada iteración, el candidato debe ser el máximo de los elementos examinados hasta ese momento.

El **tamaño de entrada**, `n`, es el número de elementos de la lista. Para empezar a analizar el tiempo, contaremos comparaciones entre un elemento y el candidato. El **espacio auxiliar** es la memoria adicional a la ocupada por la entrada.

### Puesta en común: de contar operaciones a expresar el coste

Con un recorrido de los índices 1 a `n - 1`, hacemos exactamente `n - 1` comparaciones con el candidato. Bajo el modelo habitual en el que cada comparación y acceso cuesta tiempo constante, el tiempo es **Θ(n)**: crece linealmente con la longitud.

Si solo mantenemos el candidato y el índice, el espacio auxiliar es **O(1)**: no crece con la longitud de la lista. Estas notaciones describen el crecimiento del coste, no segundos ni bytes concretos.

### Herramientas para las ampliaciones

Una función puede devolver dos valores mediante `return valor, posicion`: en realidad devuelve una **tupla**, una colección ordenada que no permite reemplazar sus elementos. Podemos desempaquetarla con `valor, posicion = funcion(...)`.

**Ejemplo resuelto de devolución doble**

````python
def doble_y_triple(numero):
    return 2 * numero, 3 * numero

doble, triple = doble_y_triple(5)
print("Doble:", doble)
print("Triple:", triple)
````
