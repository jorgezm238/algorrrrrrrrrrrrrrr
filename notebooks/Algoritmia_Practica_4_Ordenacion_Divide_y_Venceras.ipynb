# 5. Quicksort: la división depende del pivote

Quicksort no divide necesariamente la lista exactamente por la mitad.

Selecciona un **pivote** y separa los elementos en:

- menores;
- iguales;
- mayores.

Vamos a implementar una versión pedagógica.


````python
def elegir_pivote(datos, estrategia="medio"):
    if estrategia == "primero":
        return datos[0]

    if estrategia == "medio":
        return datos[len(datos) // 2]

    if estrategia == "aleatorio":
        return random.choice(datos)

    raise ValueError("Estrategia de pivote no reconocida")
````


# 6. Quicksort instrumentado: ¿qué hace el pivote?

Para el experimento comparativo vamos a usar una versión ya instrumentada.

Devuelve:

```python
(lista_ordenada, comparaciones, llamadas, profundidad_maxima)
```

Lee el código e intenta identificar dónde se cuenta cada cosa.


````python
def quicksort_contando(datos, key=lambda x: x, estrategia="medio", nivel=0):
    if len(datos) <= 1:
        return datos.copy(), 0, 1, nivel

    pivote = elegir_pivote(datos, estrategia)
    valor_pivote = key(pivote)

    menores = []
    iguales = []
    mayores = []
    comparaciones = 0

    for x in datos:
        comparaciones += 1

        if key(x) < valor_pivote:
            menores.append(x)
        elif key(x) > valor_pivote:
            mayores.append(x)
        else:
            iguales.append(x)

    izq, c_izq, l_izq, p_izq = quicksort_contando(
        menores, key, estrategia, nivel + 1
    )
    der, c_der, l_der, p_der = quicksort_contando(
        mayores, key, estrategia, nivel + 1
    )

    return (
        izq + iguales + der,
        comparaciones + c_izq + c_der,
        1 + l_izq + l_der,
        max(nivel, p_izq, p_der),
    )
````

# 7. Código auxiliar para el experimento

Para que el resto de la práctica no dependa de que alguna función anterior tenga todavía un error, utilizaremos a partir de aquí unas versiones instrumentadas proporcionadas.

No son nuevas ideas: implementan exactamente los algoritmos que acabamos de trabajar.


````python
def _bubble_ref(datos, key=lambda x: x):
    datos = datos.copy()
    comparaciones = 0
    intercambios = 0
    n = len(datos)

    for i in range(n - 1):
        cambiado = False

        for j in range(0, n - 1 - i):
            comparaciones += 1

            if key(datos[j]) > key(datos[j + 1]):
                datos[j], datos[j + 1] = datos[j + 1], datos[j]
                intercambios += 1
                cambiado = True

        if not cambiado:
            break

    return datos, comparaciones, intercambios


def _merge_ref(a, b, key=lambda x: x):
    i = 0
    j = 0
    resultado = []
    comparaciones = 0

    while i < len(a) and j < len(b):
        comparaciones += 1

        if key(a[i]) <= key(b[j]):
            resultado.append(a[i])
            i += 1
        else:
            resultado.append(b[j])
            j += 1

    resultado.extend(a[i:])
    resultado.extend(b[j:])

    return resultado, comparaciones


def _merge_sort_ref(datos, key=lambda x: x):
    if len(datos) <= 1:
        return datos.copy(), 0

    medio = len(datos) // 2

    izq, c_izq = _merge_sort_ref(datos[:medio], key)
    der, c_der = _merge_sort_ref(datos[medio:], key)

    combinado, c_merge = _merge_ref(izq, der, key)

    return combinado, c_izq + c_der + c_merge
````


# 8. Distintos tipos de entrada

Un mismo algoritmo puede comportarse de forma diferente dependiendo de los datos.

Generaremos cinco tipos:

- aleatoria;
- ordenada;
- inversa;
- casi ordenada;
- muchos repetidos.


````python
def generar_entrada(n, tipo):
    if tipo == "aleatoria":
        return random.sample(range(n * 10), n)

    if tipo == "ordenada":
        return list(range(n))

    if tipo == "inversa":
        return list(range(n - 1, -1, -1))

    if tipo == "casi ordenada":
        datos = list(range(n))
        cambios = max(1, n // 20)

        for _ in range(cambios):
            i = random.randrange(n)
            j = random.randrange(n)
            datos[i], datos[j] = datos[j], datos[i]

        return datos

    if tipo == "repetidos":
        return [random.randrange(max(2, n // 10)) for _ in range(n)]

    raise ValueError("Tipo no reconocido")
````

# 11. Experimento 2: tiempo real

Ahora sí mediremos tiempo.

Recuerda:

> Una medición concreta depende de Python, del ordenador y de muchos detalles de implementación.

La complejidad no se deduce de una única medición.

Para no hacer esperar demasiado a Colab, Bubble Sort se ejecutará solo con tamaños moderados.


````python
def medir(funcion, datos, repeticiones=3):
    tiempos = []

    for _ in range(repeticiones):
        inicio = perf_counter()
        funcion(datos)
        fin = perf_counter()
        tiempos.append(fin - inicio)

    return min(tiempos)
````

````python
filas_tiempo = []

for n in [100, 300, 600, 1000]:
    datos = generar_entrada(n, "aleatoria")

    t_bubble = medir(lambda x: _bubble_ref(x), datos)
    t_merge = medir(lambda x: _merge_sort_ref(x), datos)
    t_quick = medir(
        lambda x: quicksort_contando(x, estrategia="medio"),
        datos,
    )

    filas_tiempo.append({
        "n": n,
        "Bubble (s)": t_bubble,
        "Merge (s)": t_merge,
        "Quick (s)": t_quick,
    })

df_tiempos = pd.DataFrame(filas_tiempo)
df_tiempos
````
